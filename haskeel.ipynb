```python
%%writefile Assignment.hs

data Learner = Learner Int String [Int]

learners :: [Learner]
learners =
    [ Learner 101 "Ali" [85, 90, 95]
    , Learner 102 "Siti" [80, 80, 80]
    , Learner 103 "Ahmad" [70, 75, 72]
    , Learner 104 "Aina" [90, 85, 95]
    , Learner 105 "John" [75, 80, 85]
    ]

averageScore :: Learner -> Int
averageScore (Learner _ _ scores) =
    sum scores `div` length scores

highAchievers :: [Learner] -> [Learner]
highAchievers =
    filter (\learner -> averageScore learner >= 80)

topLearner :: [Learner] -> Learner
topLearner (x:xs) =
    foldl compareLearner x xs
    where
        compareLearner a b
            | averageScore a > averageScore b = a
            | averageScore a < averageScore b = b
            | otherwise = smallerID a b

        smallerID a@(Learner id1 _ _) b@(Learner id2 _ _)
            | id1 < id2 = a
            | otherwise = b

displayLearner :: Learner -> String
displayLearner (Learner id name scores) =
    show id ++ " " ++ name ++ " Average = "
    ++ show (averageScore (Learner id name scores))

main :: IO ()
main = do

    -- Normal test case
    let normal = Learner 106 "Mira" [70, 75, 80]

    -- Boundary test case
    let boundary = Learner 107 "Danish" [80, 80, 80]

    -- Tie test case
    let tie =
            [ Learner 108 "Sara" [90, 90, 90]
            , Learner 109 "Haziq" [90, 90, 90]
            ]

    putStrLn "Normal Test Case:"
    putStrLn (displayLearner normal)

    putStrLn "\nBoundary Test Case:"
    putStrLn (displayLearner boundary)

    putStrLn "\nTie Test Case:"
    putStrLn (displayLearner (topLearner tie))
```
